```python
def word_frequencies(filename):
    with open(filename) as f:
        words = f.read().upper().split()
        return {word: words.count(word) for word in words}
```

```python
def word_frequencies_set(filename):
    with open(filename) as f:
        words = f.read().upper().split()
        return {word: words.count(word) for word in words}
```

```python
# @title Бенчмарк: set() vs без set()
!wget -nc -q 'https://www.gutenberg.org/files/2776/2776-0.txt'

# Добавим в функцию set()
def word_frequencies_set(filename):
    with open(filename) as f:
        words = f.read().upper().split()
        return {word: words.count(word) for word in set(words)}

import time

start = time.time()
word_frequencies('2776-0.txt')
print("no set():", time.time() - start)

start = time.time()
word_frequencies_set('2776-0.txt')
print("set():", time.time() - start)

```
