```python
import json

def export_stats(*files_names):
    s = []
    for fn in files_names:
        with open(fn, "r", encoding="utf-8") as f:
            s += json.load(f)

    users_list = list(set(x['user'] for x in s))

    users = [{'user' : u,
                     'comments': len([x['user'] for x in s if x['user'] == u]),
                     'likes': sum([x['likes'] for x in s if x['user'] == u])}
                    for u in users_list]

    sorted_users = sorted(users, key=lambda x: (-x['likes'], x['user']))

    final_results = {
        "total_comments": len(s),
        "total_likes": sum(x['likes'] for x in s),
        "total_users": len(set(x['user'] for x in s)),
        "total_categories": len(files_names),
        "users": sorted_users
        }

    with open('stats.json', 'w', encoding='utf-8') as f:
        json.dump(final_results,
                  f, indent=2, ensure_ascii=False)
```

```python
# @title Короткая проверка
!wget -nc -q 'https://raw.githubusercontent.com/artemegorof/modules_and_files/refs/heads/main/module_04/lesson_03/step_17/python_comments.json'
!wget -nc -q 'https://raw.githubusercontent.com/artemegorof/modules_and_files/refs/heads/main/module_04/lesson_03/step_17/javascript_comments.json'

export_stats("python_comments.json", "javascript_comments.json")

with open('stats.json', 'r', encoding='utf-8') as f:
    print(f.read())

# {
#   "total_comments": 5,
#   "total_likes": 52,
#   "total_users": 3,
#   "total_categories": 2,
#   "users": [
#     {
#       "user": "artem",
#       "comments": 2,
#       "likes": 23
#     },
#     {
#       "user": "maria",
#       "comments": 2,
#       "likes": 22
#     },
#     {
#       "user": "john",
#       "comments": 1,
#       "likes": 7
#     }
#   ]
# }
```
