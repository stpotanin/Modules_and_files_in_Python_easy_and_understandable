```python
# @title 📜 Вариант 1. Сериализация в строку.
import json

def save_profile(profile):
    json_text = json.dumps(profile, ensure_ascii=False, indent=2)
    with open('profile.json', 'w', encoding='utf-8') as p:
        p.write(json_text)
```

```python
# @title 💾 Вариант 2. Сериализация в файл.
import json

def save_profile(profile):
    with open('profile.json', 'w', encoding='utf-8') as p:
        json.dump(profile, p, ensure_ascii=False, indent=2)
```

```python
# @title Проверка
content = {
    "name": "Артем",
    "age": 35,
    "hobbies": ["теннис", "чтение", "программирование"],
    "is_student": False
}

save_profile(content)

with open('profile.json', 'r', encoding='utf-8') as f:
    print(f.read())

# {
#   "name": "Артем",
#   "age": 35,
#   "hobbies": [
#     "теннис",
#     "чтение",
#     "программирование"
#   ],
#   "is_student": false
# }
print('------------------------------')

profile = {
    "name": "Мария",
    "age": 28,
    "city": "Санкт-Петербург",
    "languages": ["Python", "JavaScript"],
    "subscribed": None
}

save_profile(profile)

with open('profile.json', 'r', encoding='utf-8') as f:
    print(f.read())

# {
#   "name": "Мария",
#   "age": 28,
#   "city": "Санкт-Петербург",
#   "languages": [
#     "Python",
#     "JavaScript"
#   ],
#   "subscribed": null
# }
```
