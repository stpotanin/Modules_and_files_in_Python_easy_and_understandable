```python
with open('numbers.txt', 'r') as f:
    lines = f.readlines()
    first = len([x for x in lines if len(x.strip()) == 3])
    second = sum([int(x) for x in lines if len(x.strip()) == 2])

print(first, second, sep=',')
```
