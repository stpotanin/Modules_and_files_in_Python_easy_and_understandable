```python
import json

def export_poll_results(*args):
    results = {}
    # Суммируем голоса
    for arg in args:
        for k, v in arg.items():
            results[k] = results.get(k, 0) + v

    # Группируем одинаковые голоса (ключи - голоса, значения - списки из языков)
    g = {}
    for k, v in results.items():
        g[v] = g.get(v, []) + [k]

    # Собираем в  список из словарей по шаблону задания
    final_results = [{"languages":sorted(names),
                      "votes":votes,
                      "percent":votes * 100 / sum([int(k) * len(v)
                      for k, v in g.items()])} for votes, names in g.items()]
    final_results.sort(key=lambda x: -x['votes'])

    with open('results.json', 'w', encoding='utf-8') as f:
            json.dump(final_results, f, indent=2, ensure_ascii=False)
```

```python
# @title Короткая проверка
export_poll_results(
    {"Python": 21, "JavaScript": 25, "C++": 10},
    {"Python": 12, "C++": 15, "Java": 17}
)

with open('results.json', 'r', encoding='utf-8') as f:
    print(f.read())

# [
#   {
#     "languages": [
#       "Python"
#     ],
#     "votes": 33,
#     "percent": 33.0
#   },
#   {
#     "languages": [
#       "C++",
#       "JavaScript"
#     ],
#     "votes": 25,
#     "percent": 25.0
#   },
#   {
#     "languages": [
#       "Java"
#     ],
#     "votes": 17,
#     "percent": 17.0
#   }
# ]
```
