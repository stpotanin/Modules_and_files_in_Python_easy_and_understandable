## Финал Большого голосования
После успеха первого опроса сообщество **«Код без кофеина»** решило провести расширенное голосование: теперь каждый отдел компании — разработчики, тестировщики, аналитики и даже HR — выбрал свой любимый язык программирования.

Теперь нужно объединить все результаты и определить **абсолютного победителя**.

Твоя задача — написать функцию **`export_poll_results`**, которая принимает **любое количество словарей** с результатами голосования от разных отделов. Например:
```
{
    "Python": 38,
    "JavaScript": 25
},
{
    "Python": 12,
    "C++": 15,
    "Java": 17
}
```
                  
Все голоса по одинаковым языкам нужно **суммировать**, а итоговые данные сохранить в файл **`results.txt`** в таком виде:
```
Итоги общего голосования:
1. Python — 50 голосов
2. JavaScript — 25 голосов
3. Java — 17 голосов
4. C++ — 15 голосов
```
                  
Голоса сортируются **по убыванию количества голосов** — от победителя к аутсайдерам.

### 📌 Правила и требования
1. Функция может принимать любое количество словарей.
2. Все значения с одинаковыми ключами суммируются.
3. Результаты нужно **отсортировать по убыванию голосов** — от самого популярного языка к менее популярным.
4. Если несколько языков набрали **одинаковое количество голосов**, их следует вывести **на одной строке через запятую**, в **алфавитном порядке**, например:
```
2. C++, Java — 20 голосов
```
                  
5. Имя файла всегда одинаковое — `results.txt`.

````python
def export_poll_results(*args):
    results = {}
    # Суммируем голоса
    for arg in args:
        for k, v in arg.items():
            results[k] = results.get(k, 0) + v

    # Группируем одинаковые голоса (ключи - голоса, значения - списки из языков)
    g = {}
    for k, v in results.items():
        g[v] = g.get(v, []) + [k]

    # Объединяем списки из языков в строки в алф. порядке
    final_results = {', '.join(sorted(names)): votes for votes, names in g.items()}

    with open('results.txt', 'w', encoding='utf-8') as f:
        f.write('Итоги общего голосования:\n')
        for i, ((name, votes)) in enumerate(sorted(final_results.items(), key=lambda x: -x[1]), 1):
            f.write(f'{i}. {name} — {votes} голосов\n')
````

````python
# @title Проверка
export_poll_results(
    {"Python": 38, "JavaScript": 25},
    {"Python": 12, "C++": 15, "Java": 17}
)

with open('results.txt', 'r', encoding='utf-8') as f:
    print(f.read())

# Итоги общего голосования:
# 1. Python — 50 голосов
# 2. JavaScript — 25 голосов
# 3. Java — 17 голосов
# 4. C++ — 15 голосов

export_poll_results(
    {"Python": 12, "C++": 8, "Go": 9, "Swift": 6},
    {"Rust": 12, "Kotlin": 8, "Ruby": 6},
    {"Python": 3, "Go": 6, "Swift": 4, "Rust": 1},
    {"Kotlin": 2, "Ruby": 4, "JavaScript": 7}
)

with open('results.txt', 'r', encoding='utf-8') as f:
    print(f.read())

# Итоги общего голосования:
# 1. Go, Python — 15 голосов
# 2. Rust — 13 голосов
# 3. Kotlin, Ruby, Swift — 10 голосов
# 4. C++ — 8 голосов
# 5. JavaScript — 7 голосов

devs = {"Python": 30, "Java": 10, "Rust": 25, "C#": 5}
test = {"Python": 20, "Go": 15, "JavaScript": 10, "C++": 7}
analyst = {"Kotlin": 12, "JavaScript": 5, "Rust": 10}
export_poll_results(devs, test, analyst)

with open('results.txt', 'r', encoding='utf-8') as f:
    print(f.read())

# Итоги общего голосования:
# 1. Python — 50 голосов
# 2. Rust — 35 голосов
# 3. Go, JavaScript — 15 голосов
# 4. Kotlin — 12 голосов
# 5. Java — 10 голосов
# 6. C++ — 7 голосов
# 7. C# — 5 голосов

export_poll_results(
    {"Python": 20, "Java": 15, "C++": 10, "Go": 5},
    {"Rust": 25, "Java": 10, "Kotlin": 20},
    {"C": 15, "JavaScript": 10, "Go": 15, "Kotlin": 5},
    {"Swift": 12, "Python": 10, "Go": 5}
)

with open('results.txt', 'r', encoding='utf-8') as f:
    print(f.read())

# Итоги общего голосования:
# 1. Python — 30 голосов
# 2. Go, Java, Kotlin, Rust — 25 голосов
# 3. C — 15 голосов
# 4. Swift — 12 голосов
# 5. C++, JavaScript — 10 голосов
````
