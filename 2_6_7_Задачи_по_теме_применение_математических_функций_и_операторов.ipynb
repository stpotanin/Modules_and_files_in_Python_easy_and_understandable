Создайте `pd.Series`, содержащий значения `1.0, 2.0, 3.0`. Примените к каждому элементу функцию синуса и сохраните результат в новый `pd.Series`. Распечатайте обе Series.

```python
import pandas as pd
from numpy import sin

a = pd.Series([1.0, 2.0, 3.0])
print(a)

b = sin(a)
print(b)
```
