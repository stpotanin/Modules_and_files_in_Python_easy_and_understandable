# Рейтинги водителей

```python
import xml.etree.ElementTree as ET

def export_driver_ratings(filename, ratings):
    d = {}
    for driver in ratings:
        d[driver["driver"]] = d.get(driver["driver"], [0, 0])
        d[driver["driver"]][0] += driver["rating"]
        d[driver["driver"]][1] += 1
    root = ET.Element("drivers")
    for driver in d:
        dr = ET.SubElement(root, "driver", name=driver)
        ET.SubElement(dr, "avg_rating").text = str(round(d[driver][0] / d[driver][1], 1))
        ET.SubElement(dr, "reviews").text = str(d[driver][1])

    tree = ET.ElementTree(root)
    ET.indent(tree)
    tree.write(filename, encoding="utf-8")
```

```python
# @title Краткая проверка
drivers = [
    {"driver": "Batman", "rating": 5},
    {"driver": "Batman", "rating": 4},
    {"driver": "Spider-Man", "rating": 3}
]

export_driver_ratings('drivers1.xml', drivers)

with open('drivers1.xml', 'r', encoding='utf-8') as f:
    print(f.read())
# <drivers>
#   <driver name="Batman">
#     <avg_rating>4.5</avg_rating>
#     <reviews>2</reviews>
#   </driver>
#   <driver name="Spider-Man">
#     <avg_rating>3.0</avg_rating>
#     <reviews>1</reviews>
#   </driver>
# </drivers>
```
