```python
def dict_to_xml(data):
    xml = '<Students>\n'
    for d in data['Students']:
        xml += '  <Student>\n'
        for k, v in d.items():
            xml += f'    <{k}>{v}</{k}>\n'
        xml += '  </Student>\n'
    xml += '</Students>\n'

    with open("students.xml", "w", encoding="utf-8") as f:
        f.write(xml)
```
