```python
def count_lines_in_file(text_file):
    with open(text_file, 'r') as f:
        x = f.read().split('\n')
        x = x[:-1] if x[-1] == '' else x
        return len(x)

print(count_lines_in_file('dark_knight.txt')) # 11
```

```python
count_lines_in_file = lambda text_file: len(open(text_file, 'r').readlines())

print(count_lines_in_file('dark_knight.txt'))  # 11




```
