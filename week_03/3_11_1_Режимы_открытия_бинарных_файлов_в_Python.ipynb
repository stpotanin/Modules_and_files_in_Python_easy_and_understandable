```python
# @title 2. Что увидим на экране после запуска следующего кода, если файл image.png существует?
with open('image.png', 'rb') as f:
    content = f.read()
    print(type(content))
# <class 'bytes'>
```

```python
# @title 4. Что произойдёт при выполнении этого кода?
with open('file.bin', 'wb') as f:
    f.write("Привет, мир!")
# TypeError: a bytes-like object is required, not 'str'
```

```python
# @title 5. Что увидим на экране после запуска следующего кода?
text = "Данные"
data = text.encode('utf-8')

with open('info.bin', 'wb') as f:
    f.write(data)

with open('info.bin', 'rb') as f:
    result = f.read()

print(result.decode('utf-8'))
```

```python
# @title 7. Что произойдет при выполнении следующего кода?
with open("logo.png", "rb", encoding="utf-8") as f:
    data = f.read()
# ValueError: binary mode doesn't take an encoding argument
```

```python
# @title 8. Что выведет следующий код?
text = "Привет"
data = text.encode("utf-8")

with open("test.bin", "wb") as f:
    f.write(data)

with open("test.bin", "rb") as f:
    content = f.read()
    print(content)
```

```python
# @title 9. Что увидим на экране после запуска следующего кода?
with open('sample.bin', 'wb+') as f:
    f.write(b'Python\n')
    f.write(b'Is cool\n')
    f.seek(0)
    f.write(b'Hi')

    f.seek(0)
    print(f.read())
```

```python
# @title 10. Как корректно записать строку с русскими символами в бинарный файл?
with open("privet.bin", "wb") as f:
    f.write("Привет!".encode('utf-8'))
```
