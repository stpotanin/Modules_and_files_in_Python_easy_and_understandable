```python
def count_destroyed_ships(ships_file, shots_file):
    with open (ships_file) as ships, open (shots_file) as shots:
        hits = set([x.strip() for x in shots if x.strip()])
        goals = set([x.strip() for x in ships if x.strip()])
        return sum([all(coord in hits for coord in x.split()) for x in goals])
```
