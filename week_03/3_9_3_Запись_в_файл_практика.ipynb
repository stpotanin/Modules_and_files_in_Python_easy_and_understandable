## Очистка от копипасты
Каждый день сотрудники офиса сохраняют одни и те же отчёты, копируя текст из шаблонов или прошлых документов. В результате, в файле оказываются одинаковые строки по несколько раз — сплошной беспорядок!

Ваша задача — **написать функцию** `remove_duplicate_lines(filename: str, output_filename: str)`, которая:

     ☑️  прочитает содержимое файла `filename`,

     ☑️  удалит повторяющиеся строки, сохранив только первое появление каждой;

     ☑️  запишет результат в новый файл `output_filename`,

     ☑️  при этом порядок строк должен сохраниться таким, как в оригинале.

### Примечания к задаче:
     🔘  Сравнение строк — **регистрозависимое**. То есть `Hello` и `hello` — это разные строки;

     🔘  Пустые строки, если они встречаются, должны не учитываться;

     🔘  Сравнение строк выполняется по их содержимому, за исключением знака переноса строки (см. следующий пункт);

     🔘  Символ перевода строки `\n` при обработке следует удалить , **но никакие другие пробелы** и знаки пунктуации **изменять нельзя**.

     [GitHub Repository](https://github.com/artemegorof/modules_and_files/blob/main/module_03/lesson_09/step_03/tests)

```python
def remove_duplicate_lines(filename: str, output_filename: str):
    seen_lines = set() # сет существующих строк
    with open(filename, 'r') as infile, open(output_filename, 'w') as outfile:
        for line in infile:
            cleaned_line = line.rstrip('\n')
            if not cleaned_line:
                continue # не добавляем пустые строки
            if cleaned_line not in seen_lines:
                seen_lines.add(cleaned_line)
                outfile.write(cleaned_line + '\n')
```

```python
# словарём
def remove_duplicate_lines(filename: str, output_filename: str):
    with open(filename, 'r') as f, open(output_filename, 'w') as o:
        d = {line.strip(): line.strip()+'\n' for line in f if line.strip()}
        o.writelines(d.values())
```

```python
# Екатерина Чернышева https://stepik.org/lesson/1947124/step/3?discussion=13403088&thread=solutions
def remove_duplicate_lines(filename: str, output_filename: str):
    with open(filename, 'r', encoding='utf-8') as file, open(output_filename, 'w', encoding='utf-8') as out_file:
        text = [i.strip()+'\n' for i in file.readlines()]
        text = dict.fromkeys(text)
        out_file.writelines(text)
```
