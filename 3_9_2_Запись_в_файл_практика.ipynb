## Файл наоборот
В отделе хранения данных допустили ошибку: строки в одном из текстовых файлов оказались записаны в обратном порядке. Нужно срочно исправить ситуацию и сформировать новый файл, где строки будут располагаться в правильном порядке — с конца в начало.

Ваша задача — написать функцию `reverse_lines(filename)`, которая:

     ☑️  принимает имя файла filename в качестве аргумента;

     ☑️  читает все строки из этого файла;

     ☑️  записывает их в новый файл `reversed.txt` в обратном порядке (первая становится последней, последняя — первой);

     ☑️  все символы, включая переводы строки, должны быть сохранены как есть.

Примечания к задаче:
     🔘  Пустые строки, если они встречаются, должны учитываться.

```python
# Вариант 1
def reverse_lines(filename):
    with open(filename, 'r') as f:
        lines = f.readlines()
    with open('reversed.txt', 'w') as f:
        [f.write(line.strip() + '\n') for line in reversed(lines)]
```

```python
# Вариант 2
def reverse_lines(filename):
    with open(filename, 'r') as f, open('reversed.txt', 'w') as s:
        [s.write(line.strip() + '\n') for line in f.readlines()[::-1]]
```

```python
# Александр Павлов https://stepik.org/lesson/1947124/step/2?discussion=13364091&thread=solutions
def reverse_lines(filename):
    with (open(filename, encoding='utf-8') as file_in,
          open('reversed.txt', 'w', encoding='utf-8') as file_out):
        file_out.writelines('\n'.join([*map(str.strip, file_in.readlines())][::-1]))
```
