# Покупатели и категории покупок

```python
import csv

def export_category_summary(filename):
    with open(filename, encoding='utf-8') as f:
        data = csv.reader(f)
        next(data)
        d = {}

        for row in data:
            c = d.setdefault(row[2], [0, set()]) # ссылка на значение ключа
            c[0] += int(row[3])  # сумма продажи, меняем по ссылке
            c[1].add(row[0])     # клиент, меняем по ссылке

        res = {}
        for k, v in d.items():
            res[k] = [v[0], len(v[1])]  # меняем список на его длину

    with open('category_summary.csv', 'w', encoding='utf-8', newline='') as f:
        csv_writer = csv.writer(f)
        csv_writer.writerow(['category', 'total_sales', 'unique_customers'])

        for k, v in sorted(res.items()):
            csv_writer.writerow([k, v[0], v[1]])
```
