### Уборка архива
В архиве отдела находятся старые файлы, в которых между строками затесались пустые строки — возможно, по ошибке, а возможно, от скуки стажёра 🤷‍♂️.

Ваша задача — написать функцию `clean_file(filename)`, которая:

* читает содержимое указанного файла,
* удаляет все полностью пустые строки,
* и сохраняет результат в новый файл cleaned.txt.

Примечание по решению:

* Нужно сохранить порядок строк.
* Удаляются только те строки, в которых нет символов вообще (включая пробелы).
* В файле может быть как одна строка, так и десятки — будьте готовы ко всему.

```python
def clean_file(filename):
    with open(filename, 'r', encoding='utf-8') as f, open('cleaned.txt', 'w', encoding='utf-8') as c:
        c.writelines([x for x in f.readlines() if x.strip()])
```
