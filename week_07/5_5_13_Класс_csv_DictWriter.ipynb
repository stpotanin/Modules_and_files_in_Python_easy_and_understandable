# Экспорт нескольких колонок каталога

```python
import csv

def export_columns(products, filename, columns):
    col_list = [x for x in columns if x in products[0].keys()]
    if not col_list:
        return print('Ни одна из указанных колонок не найдена.')

    with open(filename, 'w', encoding='utf-8', newline='') as f:
        writer = csv.DictWriter(f, columns, restval='N/A')
        writer.writeheader()
        writer.writerows([{col_name: x[col_name] for col_name in col_list} for x in products])
```

```python
# @title Короткая проверка
catalog = [
    {"product": "Чай", "category": "Напитки", "price": 100},
    {"product": "Кофе", "category": "Напитки", "price": 200},
    {"product": "Печенье", "category": "Сладости", "price": 150}
]

export_columns(catalog, "selected.csv", ["product", "price"])

with open('selected.csv', 'r', encoding='utf-8') as f:
    print(f.read())

# product,price
# Чай,100
# Кофе,200
# Печенье,150
```
