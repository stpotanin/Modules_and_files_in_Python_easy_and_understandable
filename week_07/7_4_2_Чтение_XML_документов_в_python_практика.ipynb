# Дочерние теги корневого элемента

```python
import xml.etree.ElementTree as ET

def get_child_tags(filename):
    tree = ET.parse(filename)
    root = tree.getroot()
    return [x.tag for x in root]
```

```python
import xml.etree.ElementTree as ET

get_child_tags = lambda filename: [x.tag for x in ET.parse(filename).getroot()]
```
