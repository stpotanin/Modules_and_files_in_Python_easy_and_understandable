# Расширенное чтение CSV-файла

```python
import csv

def read_csv(filename, show_header=True, limit=None):
    with open(filename, encoding='utf-8') as f:
        res =  list(csv.reader(f))
    print(*res[int(not show_header):] if not limit else res[int(not show_header):limit+1], sep='\n')
```

```python
# @title Короткая проверка
!wget -nc -q 'https://raw.githubusercontent.com/artemegorof/modules_and_files/refs/heads/main/module_05/lesson_03/step_02/books.csv'

read_csv("books.csv", show_header=False, limit=2)

# # вызов должен распечатать на экран следующее:

# ['1', '1984', 'Джордж Оруэлл', 'Фантастика', '328', '4.8']
# ['2', 'Мастер и Маргарита', 'Михаил Булгаков', 'Роман', '480', '4.9']
```
