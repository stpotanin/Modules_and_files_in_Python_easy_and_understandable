```python
import json

def find_best_manager(filename):
    with open(filename) as f:
        data = json.load(f)
        best_manager = max(data, key=lambda m: sum(car['price'] for car in m['cars']))
        print(best_manager['manager']['first_name'],
              best_manager['manager']['last_name'],
              sum(car['price'] for car in best_manager['cars']))
```
