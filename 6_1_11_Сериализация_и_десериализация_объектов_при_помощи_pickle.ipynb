# Восстановление списка увлечений из файла

```python
import pickle

def load_hobbies(filename):
    with open(filename, 'rb') as f:
        return pickle.load(f)
```

```python
# @title Короткая проверка
import pickle

def save_hobbies(filename, hobbies):
    with open(filename, 'wb') as f:
        pickle.dump(hobbies, f)

hobbies = ["Теннис", "Программирование", "Чтение"]
save_hobbies("artem_hobbies.pkl", hobbies)


restored = load_hobbies("artem_hobbies.pkl")
print(restored)

# Ожидаемый вывод:
# ['Теннис', 'Программирование', 'Чтение']

# b'\x80\x04\x95F\x00\x00\x00\x00\x00\x00\x00]\x94(\x8c\x0c\xd0\xa2\xd0\xb5\xd0\xbd\xd0\xbd\xd0\xb8\xd1\x81\x94\x8c \xd0\x9f\xd1\x80\xd0\xbe\xd0\xb3\xd1\x80\xd0\xb0\xd0\xbc\xd0\xbc\xd0\xb8\xd1\x80\xd0\xbe\xd0\xb2\xd0\xb0\xd0\xbd\xd0\xb8\xd0\xb5\x94\x8c\x0c\xd0\xa7\xd1\x82\xd0\xb5\xd0\xbd\xd0\xb8\xd0\xb5\x94e.'
```
