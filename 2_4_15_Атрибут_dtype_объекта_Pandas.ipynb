Создайте объект Series с элементами `[True, False, True]` и типом данных `bool`. Назначьте индексы `['x', 'y', 'z']`. Выведите его на печать. Затем измените все значения на `False`. Выведите обновленные данные на печать.

```python
import pandas as pd

df = pd.Series([True, False, True],
               index=['x', 'y', 'z'],
               dtype='bool')

print(df)

df = df * False
print(df)
```
