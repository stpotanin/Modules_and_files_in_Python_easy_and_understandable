```python
print_first_two_lines = lambda f: print(''.join(list(open(f))[:2]))
```

```python
with open('chat_susan.txt', 'w') as f:
    f.write('''Hi, I can't access my account.
Did you recently change your password?
I think I forgot it again.
    ''')

with open('chat_daniel.txt', 'w') as f:
    f.write('''Good morning!
Is this the tech support chat?
Yes, how can I help you?
    ''')

with open('chat_julia.txt', 'w') as f:
    f.write('''Please cancel my order.
I placed it by mistake.
Thank you!
    ''')
```

```python
# Тест 1:
# Входные данные:
print_first_two_lines('chat_susan.txt')


# Ожидаемый вывод:
# Hi, I can't access my account.
# Did you recently change your password?
# ------------------------------

# Тест 2:
# Входные данные:
print_first_two_lines('chat_daniel.txt')


# Ожидаемый вывод:
# Good morning!
# Is this the tech support chat?
# ------------------------------

# Тест 3:
# Входные данные:
print_first_two_lines('chat_julia.txt')


# Ожидаемый вывод:
# Please cancel my order.
# I placed it by mistake.
# ------------------------------
```
