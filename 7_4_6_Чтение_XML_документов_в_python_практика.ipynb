# Курсы для каждого уровня сложности

```python
import xml.etree.ElementTree as ET

def print_courses_by_level(filename):
    root = ET.parse(filename).getroot()
    for category in root.findall("category"):
        print(f"Категория: {category.attrib['name']}")
        for level in category.findall("level"):
            print(f"  Уровень: {level.attrib['value']}")
            for course in level.findall("course"):
                # if course.attrib['active'].lower() == 'true':
                print(f"    - {course.find("title").text}")
```
