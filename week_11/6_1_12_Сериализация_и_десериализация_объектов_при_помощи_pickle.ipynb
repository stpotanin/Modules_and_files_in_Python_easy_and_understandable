# Читаем функцию из pickle-файла

```python
import pickle

def load_and_describe_function(filename):
    with open(filename, 'rb') as f:
        func = pickle.load(f)
    print(f"Имя функции: {func.__name__}")
    if func.__doc__:
        print(f"Описание: {func.__doc__.strip()}")
    else:
        print("Описание: Нет описания (docstring)")
    return func
```

```python
# @title Проверка
!wget -nc -q 'https://github.com/artemegorof/modules_and_files/raw/refs/heads/main/module_06/lesson_01/step_12/greet.pkl'
!wget -nc -q 'https://github.com/artemegorof/modules_and_files/raw/refs/heads/main/module_06/lesson_01/step_12/factorial.pkl'
!wget -nc -q 'https://github.com/artemegorof/modules_and_files/raw/refs/heads/main/module_06/lesson_01/step_12/multiplier.pkl'
!wget -nc -q 'https://github.com/artemegorof/modules_and_files/raw/refs/heads/main/module_06/lesson_01/step_12/add.pkl'

load_and_describe_function("greet.pkl")
# Имя функции: greet
# Описание: Функция приветствует пользователя.
print()

load_and_describe_function("factorial.pkl")
# Имя функции: factorial
# Описание: Вычисляет факториал числа n. Использует рекурсию.
print()

load_and_describe_function("multiplier.pkl")
# Имя функции: multiplier
# Описание: Возвращает функцию умножения на k.
print()

load_and_describe_function("add.pkl")
# Имя функции: add
# Описание: Нет описания (docstring)
```
