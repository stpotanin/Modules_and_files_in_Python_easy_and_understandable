```python
def dict_to_xml(data):
    xml = '<Students>\n'
    for d in data['Students']:
        xml += '  <Student>\n'
        for k, v in d.items():
            if type(v) is list:
                xml += f'    <{k}>\n'
                for vv in v:
                    xml += f'      <{k[:-1]}>{vv}</{k[:-1]}>\n'
                xml += f'    </{k}>\n'
            else:
                xml += f'    <{k}>{v}</{k}>\n'
        xml += '  </Student>\n'
    xml += '</Students>'

    with open("students.xml", "w", encoding="utf-8") as f:
        f.write(xml)
```
