```python
def find_words_ending_with_eya(filename):
    with open (filename) as f:
        words = sorted(set([x for x in f.read().upper().split() \
                       if x[-2:] == 'ЕЯ']),
                       key = lambda k: (len(k), k))
    [print(w) for w in words]
```

```python
def find_words_ending_with_eya(filename):
    with open (filename) as f:
        words = sorted(set([x for x in f.read().upper().split() \
                       if x.endswith('ЕЯ')]),
                       key = lambda k: (len(k), k))
    [print(w) for w in words]
```
