```python
def dict_to_xml(data):
    xml = '<Students>\n'
    # уровень 1
    for d in data['Students']:

        xml += '  <Student>\n'
        # уровень 2
        for k, v in d.items():
            if type(v) is list:
                xml += f'    <{k}>\n'
                # уровень 3
                for vv in v:
                    if type(vv) is list:
                        for vvv in vv:
                             xml += f'      <{k[:-1]}>{vvv}</{k[:-1]}>\n'
                    elif type(vv) is dict:
                        for kkk, vvv in vv.items():
                             xml += f'      <{kkk}>{vvv}</{kkk}>\n'
                    else:
                        xml += f'      <{k[:-1]}>{vv}</{k[:-1]}>\n'
                xml += f'    </{k}>\n'

            elif type(v) is dict:
                xml += f'    <{k}>\n'
                # уровень 3
                for kk, vv in v.items():
                    if type(vv) is list:
                        for vvv in vv:
                            xml += f'        <{vv[:-1]}>{vvv}</{vv[:-1]}>\n'
                    elif type(vv) is dict:
                        for kkk, vvv in vv.items():
                            xml += f'        <{kkk}>{vvv}</{kkk}>\n'
                    else:
                        xml += f'      <{kk}>{vv}</{kk}>\n'
                xml += f'    </{k}>\n'
            else:
                xml += f'    <{k}>{v}</{k}>\n'
        xml += '  </Student>\n'
    xml += '</Students>\n'

    with open("students.xml", "w", encoding="utf-8") as f:
        f.write(xml)
```
