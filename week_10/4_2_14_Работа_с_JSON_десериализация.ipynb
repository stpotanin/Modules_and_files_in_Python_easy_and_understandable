```python
# @title Циклом
import json

def count_them(filename):
    id_group_max = 0
    count_max = 0

    with open(filename) as f:
        data = json.load(f)

        for group in data:
            id_group = group.get('id_group')
            people = group.get('people')

            count = len([p for p in people if p.get('gender') == 'Female' and p.get('year') > 1977])

            if count > count_max:
                count_max = count
                id_group_max = id_group

        print(id_group_max, count_max)
```

```python
# @title Функционально
import json


def count_them(filename):
    count = lambda people: len(['' for p in people if p.get('gender') == 'Female' and p.get('year') > 1977])
    with open(filename) as f:
        data = json.load(f)
        max_group = max(data, key=lambda group: count(group['people']))
        print(max_group['id_group'], count(max_group['people']))
```
