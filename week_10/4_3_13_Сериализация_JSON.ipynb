```python
import json


def export_comments(comments, filename, reverse=None, key="likes"):
    if reverse is not None:
        comments = sorted(
            comments,
            key=lambda x: x.get(key, 0),
            reverse=bool(reverse)
        )

    with open(filename, 'w', encoding='utf-8') as f:
        json.dump(
            comments,
            f,
            ensure_ascii=False,
            indent=2
        )
```
