```python
# @title 1. Программа списка дел
task = input("Введите новую задачу: ")

with open('todo.txt', 'a+', encoding='utf-8') as f:
    f.write(task + '\n')
    f.seek(0)
    print("Ваш список дел:\n" + f.read())
```

```python
# @title 3. Что произойдёт, если дважды запустить следующий код?
with open("report.txt", "a", encoding="utf-8") as f:
    f.write("Отчет готов!\n")
# К содержимому файла добавится в конец дважды строка «Отчет готов!»
```

```python
# @title 6. Какой результат будет у следующего кода при первом запуске?
with open("data.txt", "w", encoding="utf-8") as f:
    f.write("Первая строка\n")

with open("data.txt", "a", encoding="utf-8") as f:
    f.write("Вторая строка\n")

with open("data.txt", "r", encoding="utf-8") as f:
    print(f.read())
```

```python

# @title 8. Какая комбинация кода приведёт к тому, что строка текст
# будет добавляться дополнительно в файл при каждом запуске?
with open("log.txt", "a") as f:
     f.write("Hello\n")
     f.write("World\n")
```

```python
with open("log.txt", "a+") as f:
     f.write("Hello\n")
     f.write("World\n")
```

```python
# @title 9. Что напечатает следующая программа, если файл log.txt
# изначально не существует, и программа выполнится один раз?
with open("log.txt", "a+", encoding="utf-8") as f:
    f.write("Начало\nКонец")
    f.seek(0)
    content = f.read()
    print(content)
```

```python
# @title 10. Что выведет следующий код при запуске программы, если log.txt уже содержит одну строку "Запуск 1\n"?
with open("log.txt", "a+", encoding="utf-8") as f:
    f.write("Запуск 2\n")
    f.seek(0)
    print(f.read())
# Запуск 1\nЗапуск 2 или Запуск 1\\nЗапуск 2
```
