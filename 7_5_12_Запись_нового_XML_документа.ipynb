# Логирование ошибок в XML

```python
import xml.etree.ElementTree as ET

def save_errors(filename, errors):
    root = ET.Element("errors")
    for error in errors:
        name = ET.SubElement(root, 'error')
        name.text = error

    tree = ET.ElementTree(root)
    ET.indent(tree)
    tree.write(filename, encoding="utf-8")
```

```python
# @title Краткая проверка
errors = [
    "File not found",
    "Timeout error",
    "Permission denied"
]

save_errors('errors1.xml', errors)

with open('errors1.xml', 'r', encoding='utf-8') as f:
    print(f.read())
# <errors>
#   <error>File not found</error>
#   <error>Timeout error</error>
#   <error>Permission denied</error>
# </errors>
```
