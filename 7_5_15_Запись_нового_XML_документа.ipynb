# Экспорт сотрудников по отделам

```python
import xml.etree.ElementTree as ET

def export_company_structure(filename, departments):
    root = ET.Element("company")
    for k, v in departments.items():
        department = ET.SubElement(root, "department")
        department.attrib["name"] = k
        for e in v:
            employee = ET.SubElement(department, "employee")
            employee.text = e

    tree = ET.ElementTree(root)
    ET.indent(tree)
    tree.write(filename, encoding="utf-8")
```

```python
# @title Краткая проверка
departments = {
    "IT": ["Anna", "Leo"],
    "HR": ["Maria"]
}

export_company_structure('company1.xml', departments)

with open('company1.xml', 'r', encoding='utf-8') as f:
    print(f.read())
# <company>
#   <department name="IT">
#     <employee>Anna</employee>
#     <employee>Leo</employee>
#   </department>
#   <department name="HR">
#     <employee>Maria</employee>
#   </department>
# </company>
```
