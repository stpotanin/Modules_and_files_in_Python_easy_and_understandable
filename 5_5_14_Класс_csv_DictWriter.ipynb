# Экспорт каталога товаров с заменой названий полей

```python
import csv, json

def export_catalog(data, mapping_filename, output_filename):
    with open(mapping_filename, encoding='utf-8') as fm, open(output_filename, 'w', encoding='utf-8', newline='') as f:
        translate = json.load(fm)
        fieldnames = list(v for k, v in translate.items()) + [k for k in data[0].keys() if k not in translate.keys()]

        writer = csv.DictWriter(f, fieldnames)
        writer.writeheader()
        writer.writerows([{translate.get(k, k): v for k, v in d.items()} for d in data])
```
