# Каршеринг: средняя поездка

```python
import xml.etree.ElementTree as ET

def average_distance_by_city(filename):
    root = ET.parse(filename).getroot()
    d = {}
    for trip in root.findall('trip'):
        city = trip.find('city').text
        distance = trip.find('distance').text
        d[city] = d.get(city, [0, 0])
        d[city][0] += int(distance)
        d[city][1] += 1

    return {city : d[city][0] / d[city][1] for city in d}
```

```python
# @title Краткая проверка
!wget -nc -q 'https://raw.githubusercontent.com/artemegorof/modules_and_files/refs/heads/main/module_07/lesson_04/step_09/trips_1.xml'

print(average_distance_by_city("trips_1.xml"))
# {'Berlin': 10.0, 'Paris': 10.0}
```
