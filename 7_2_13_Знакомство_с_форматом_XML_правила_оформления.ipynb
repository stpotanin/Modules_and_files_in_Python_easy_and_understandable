```python
# Глубина словаря - 1 уровень
def dict_to_xml(data):
    xml = '<Student>\n'
    for k, v in data['Student'].items():
        xml += f'  <{k}>{v}</{k}>\n'
    xml += '</Student>\n'

    with open("student.xml", "w", encoding="utf-8") as f:
        f.write(xml)
```

```python
# @title Глубина словаря произвольная
def dict_to_xml(data):
    xml = ''

    def unpackage_dict(data, padding):
        nonlocal xml
        for k, v in data.items():
            if isinstance(v, dict):
                xml += ' ' * padding + f'<{k}>\n'
                unpackage_dict(v, padding + 2)
                xml += ' ' * padding + f'</{k}>\n'
            else:
                xml += ' ' * padding + f'<{k}>{v}</{k}>\n'

    unpackage_dict(data, 0)

    with open("student.xml", "w", encoding="utf-8") as f:
        f.write(xml)
```
