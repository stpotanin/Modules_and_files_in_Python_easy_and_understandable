```python
def get_unique_words(file_name, punct='''!"#$%&'()*+,-./:;<=>?@[\\]^_`{|}~'''):
    clean_text = open(file_name).read()
    for char in punct:
        clean_text = clean_text.replace(char, '')
    return sorted(set(clean_text.lower().split()))
```

```python
# Через генератор множеств с Сергеем Балакиревым)
# https://stepik.org/lesson/1935163/step/14?discussion=13067574&thread=solutions
def get_unique_words(file_name: str) -> list:
    DEL_CHAR = '!"#$%&\'()*+,-./:;<=>?@[\\]^_`{|}~–—’'
    trans = str.maketrans("", "", DEL_CHAR)

    with open(file_name, "r", encoding="utf-8") as f:
        words = {
            word
            for line in f
            for word in line.translate(trans).lower().split()
            if word
        }

    return sorted(words)
```
