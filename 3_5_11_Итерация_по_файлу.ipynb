```python
def print_lines_with_numbers(f):
    with open(f) as file:
        for i, line in enumerate(file):
            print(f'{i+1}: {line.strip()}')
```

```python
print_lines_with_numbers = lambda f: [print(f'{i+1}: {line.strip()}') for i, line in enumerate(open(f))]
```
