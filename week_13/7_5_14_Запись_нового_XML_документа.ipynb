# Экспорт заказов

```python
import xml.etree.ElementTree as ET

def export_orders(filename, orders):
    root = ET.Element("orders")
    for d in orders:
        order = ET.SubElement(root, "order")
        order.attrib["id"] = str(d["id"])
        for k, v in d.items():
            if k != "id":
                element = ET.SubElement(order, k)
                element.text = str(v)

    tree = ET.ElementTree(root)
    ET.indent(tree)
    tree.write(filename, encoding="utf-8")
```

```python
# @title Краткая проверка
orders = [
    {"id": 101, "amount": 250, "currency": "USD"},
    {"id": 102, "amount": 400, "currency": "EUR"}
]

export_orders('orders1.xml', orders)

with open('orders1.xml', 'r', encoding='utf-8') as f:
    print(f.read())
# <orders>
#   <order id="101">
#     <amount>250</amount>
#     <currency>USD</currency>
#   </order>
#   <order id="102">
#     <amount>400</amount>
#     <currency>EUR</currency>
#   </order>
# </orders>
```
