# Экспорт одной колонки каталога

```python
import csv

def export_column(products, filename, column_name):
    if column_name in products[0]:
        with open(filename, 'w', encoding='utf-8', newline='') as f:
            writer = csv.DictWriter(f, [column_name])
            writer.writeheader()
            writer.writerows({column_name: x[column_name]} for x in products)
    else:
        print(f'Колонка {column_name} отсутствует в данных.')
```

```python
# @title Короткая проверка
import csv

catalog = [
    {"product": "Чай", "category": "Напитки", "price": 100},
    {"product": "Кофе", "category": "Напитки", "price": 200},
    {"product": "Печенье", "category": "Сладости", "price": 150}
]

export_column(catalog, 'price.csv', "price")

with open('price.csv', 'r', encoding='utf-8') as f:
    print(f.read())

# price
# 100
# 200
# 150
```
