# Средняя стоимость билета по типам пассажиров

```python
import xml.etree.ElementTree as ET

def average_fare_by_passenger_type(filename):
    root = ET.parse(filename).getroot()
    d = {}
    d['ADT'], d['CHD'] = [0, 0], [0, 0]
    for passenger in root.findall('.//passenger'):
        d[passenger.attrib['type']][0] += float(passenger.find('.//base').text) + float(passenger.find('.//tax').text)
        d[passenger.attrib['type']][1] += 1
    return {k: round(v[0] / v[1] if v[1] else 0, 2) for k, v in d.items() if v[0]}
```
