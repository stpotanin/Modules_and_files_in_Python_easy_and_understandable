# Универсальная сериализация структуры в XML

```python
import xml.etree.ElementTree as ET

def save_to_xml(data, filename):

    def save_tag(root, data):
            if isinstance(data, dict):
                for k, v in data.items():
                    tag = ET.SubElement(root, k)
                    save_tag(tag, v)
            elif isinstance(data, list):
                for v in data:
                    tag = ET.SubElement(root, root.tag[:-1])
                    save_tag(tag, v)
            else:
                root.text = str(data)

    k, v = list(data.items())[0]
    root = ET.Element(k)
    save_tag(root, v)

    tree = ET.ElementTree(root)
    ET.indent(tree)
    tree.write(filename, encoding="utf-8")
```
