# Экспорт каталога товаров в CSV

```python
import csv

def save_catalog_to_csv(catalog, filename):
    with open(filename, 'w', encoding='utf-8', newline='') as f:
        writer = csv.DictWriter(f, catalog[0].keys())
        writer.writeheader()
        writer.writerows(catalog)
```

```python
# @title Короткая проверка

catalog = [
    {"product": "Чай", "category": "Напитки", "price": 100},
    {"product": "Кофе", "category": "Напитки", "price": 200},
    {"product": "Печенье", "category": "Сладости", "price": 150}
]

save_catalog_to_csv(catalog, 'products.csv')

with open('products.csv', 'r', encoding='utf-8') as f:
    print(f.read())

# product,category,price
# Чай,Напитки,100
# Кофе,Напитки,200
# Печенье,Сладости,150
```
