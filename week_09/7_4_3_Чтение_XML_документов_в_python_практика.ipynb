# Дочерние теги заданного элемента

```python
import xml.etree.ElementTree as ET

def get_child_tags(filename, parent_tag):
    tree = ET.parse(filename)
    root = tree.getroot()
    return [xx.tag for x in root.findall(parent_tag) for xx in x]
```
