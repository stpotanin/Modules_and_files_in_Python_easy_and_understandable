# Экспорт контактов

```python
import xml.etree.ElementTree as ET

def export_contacts(filename, contacts):
    root = ET.Element("contacts")
    for d in contacts:
        contact = ET.SubElement(root, "contact")
        for k, v in d.items():
            element = ET.SubElement(contact, k)
            element.text = str(v)

    tree = ET.ElementTree(root)
    ET.indent(tree)
    tree.write(filename, encoding="utf-8")
```

```python
# @title Краткая проверка
contacts = [
    {"name": "Anna", "phone": "+374777777", "city": "Yerevan"},
    {"name": "Leo", "phone": "+374999999", "city": "Gyumri"}
]

export_contacts('contacts1.xml', contacts)

with open('contacts1.xml', 'r', encoding='utf-8') as f:
    print(f.read())
# <contacts>
#   <contact>
#     <name>Anna</name>
#     <phone>+374777777</phone>
#     <city>Yerevan</city>
#   </contact>
#   <contact>
#     <name>Leo</name>
#     <phone>+374999999</phone>
#     <city>Gyumri</city>
#   </contact>
# </contacts>
```
