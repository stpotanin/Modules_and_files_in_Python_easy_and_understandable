# Турнирная статистика


```python
import csv

def export_team_stats(filename):
    with open(filename, encoding='utf-8') as f:
        data = list(csv.reader(f))[1:]
    stat = {x[0]:[0, 0, 0, 0, 0, 0, 0] for x in data}
    for x in data:
        # хозяин:
        stat[x[0]][0] += 1
        stat[x[0]][1] += 1 if x[2] > x[3] else 0
        stat[x[0]][2] += 1 if x[2] < x[3] else 0
        stat[x[0]][3] += 1 if x[2] == x[3] else 0
        stat[x[0]][4] += int(x[2])
        stat[x[0]][5] += int(x[3])
        stat[x[0]][6] += int(x[2]) - int(x[3])
        # гость:
        stat[x[1]][0] += 1
        stat[x[1]][1] += 1 if x[2] < x[3] else 0
        stat[x[1]][2] += 1 if x[2] > x[3] else 0
        stat[x[1]][3] += 1 if x[2] == x[3] else 0
        stat[x[1]][4] += int(x[3])
        stat[x[1]][5] += int(x[2])
        stat[x[1]][6] += int(x[3]) - int(x[2])

    stat_list = ['Команда,Матчи,Победы,Поражения,Ничьи,Забито,Пропущено,Разница'] + \
                [f'{k},'+','.join(map(str, v)) for k, v in sorted(stat.items())]

    with open('team_stats.csv', 'w', encoding='utf-8') as f:
        for i in stat_list:
            f.write(f'{i}\n')
```

```python
# @title Короткая проверка
!wget -nc -q 'https://raw.githubusercontent.com/artemegorof/modules_and_files/refs/heads/main/module_05/lesson_03/step_15/match_results.csv'

export_team_stats('match_results.csv')

with open('team_stats.csv', encoding='utf-8') as f:
    print(f.read())

# Команда,Матчи,Победы,Поражения,Ничьи,Забито,Пропущено,Разница
# Динамо,2,0,0,2,3,3,0
# Зенит,2,1,1,0,4,2,2
# Спартак,2,1,0,1,4,3,1
# ЦСКА,2,0,1,1,1,4,-3
```
