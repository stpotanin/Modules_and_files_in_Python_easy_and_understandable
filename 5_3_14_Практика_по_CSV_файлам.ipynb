# Зарплатная сводка по отделам

```python
import json

def export_summary(filenames, output_filename):
    d = ['department,employees,average_salary,min_salary,max_salary']
    for filename in sorted(filenames):
        with open(filename, encoding='utf-8') as f:
            reader = json.load(f)
            if reader:
                department = filename[:-5]
                employees = len(reader)
                average_salary = round(sum([employee['salary'] for employee in reader]) / employees, 2)
                min_salary = round(float(min([employee['salary'] for employee in reader])), 2)
                max_salary = round(float(max([employee['salary'] for employee in reader])), 2)
                d.append(f'{department},{employees},{average_salary},{min_salary},{max_salary}')

    with open(output_filename, 'w', encoding='utf-8') as f:
        for line in d:
            f.write(line + '\n')
```

```python
# @title Короткая проверка
!wget -nc -q 'https://raw.githubusercontent.com/artemegorof/modules_and_files/refs/heads/main/module_05/lesson_03/step_14/it.json'
!wget -nc -q 'https://raw.githubusercontent.com/artemegorof/modules_and_files/refs/heads/main/module_05/lesson_03/step_14/hr.json'
!wget -nc -q 'https://raw.githubusercontent.com/artemegorof/modules_and_files/refs/heads/main/module_05/lesson_03/step_14/sales.json'

export_summary(['it.json', 'hr.json', 'sales.json'], "summary.csv")

with open('summary.csv', encoding='utf-8') as f:
    print(f.read())

# department,employees,average_salary,min_salary,max_salary
# hr,2,92500.0,90000.0,95000.0
# it,3,123333.33,100000.0,150000.0
# sales,3,131666.67,105000.0,180000.0
```
