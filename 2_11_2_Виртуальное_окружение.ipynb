```python
# @title  3. Если в новом окружении нет pip, устанавливаем вручную
# !curl -sS https://bootstrap.pypa.io/get-pip.py -o get-pip.py
# !/content/my_venv/bin/python get-pip.py
```

```python
# @title  4. Проверяем версии (исходные и получившиеся в виртуальном окружении)
print('Версии в основной среде:')
!python --version
!pip --version
print('Версии в виртуальном окружении:')
!/content/my_venv/bin/python --version
!/content/my_venv/bin/pip --version
```

```python
# @title  5. Устанавливаем нужные пакеты
!/content/my_venv/bin/pip install numpy==1.26.4
!/content/my_venv/bin/pip install matplotlib==3.8.2
!/content/my_venv/bin/pip install requests==2.31.0
```

```python
# @title  6. Проверка версий:
!/content/my_venv/bin/python -c "import numpy; print(f'numpy=={numpy.__version__}')"
!/content/my_venv/bin/python -c "import matplotlib; print(f'matplotlib=={matplotlib.__version__}')"
!/content/my_venv/bin/python -c "import requests; print(f'requests=={requests.__version__}')"
```
