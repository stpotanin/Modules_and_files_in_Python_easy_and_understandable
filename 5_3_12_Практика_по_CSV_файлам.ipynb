# Ученики и курсы

```python
import csv

def merge_student_courses(stud_filename, enroll_filename):
    with open(stud_filename, encoding='utf-8') as fs, open(enroll_filename, encoding='utf-8') as fe:
        stud_dict, courses_dict = {}, {}

        students = csv.reader(fs, delimiter=':')
        next(students)
        for s in students:
            stud_dict[s[0]] = s[1]

        courses = csv.reader(fe, delimiter=':')
        next(courses)
        for s in courses:
            courses_dict.setdefault(stud_dict[s[0]], [s[0]]).append(s[1])

    with open('student_courses.csv', 'w', encoding='utf-8') as f:
        f.write('name:courses\n')
        for s, _ in stud_dict.items():
            if stud_dict[s] in courses_dict:
                f.write(f'{stud_dict[s]}:{', '.join(sorted(courses_dict[stud_dict[s]])[1:])}\n')
            else:
                f.write(f'{stud_dict[s]}:Пока нет направлений обучения\n')
```

```python
# @title Вариант 2
import csv

def merge_student_courses(stud_filename, enroll_filename):
    with open(stud_filename, encoding='utf-8') as fs, open(enroll_filename, encoding='utf-8') as fe:
        students = list(csv.reader(fs, delimiter=':'))[1:]
        courses  = list(csv.reader(fe, delimiter=':')) [1:]

        s_dict = {s[0]:s[1] for s in students}
        c_dict = {s[0]:[] for s in students}

        for k, v in courses:
            c_dict[k].append(v)
        for k, v in c_dict.items():
            c_dict[k] = ', '.join(sorted(v)) if v else 'Пока нет направлений обучения'

    with open('student_courses.csv', 'w', encoding='utf-8', newline='') as f:
        writer = csv.writer(f, delimiter=':')
        writer.writerow(['name', 'courses'])
        for k, v in c_dict.items():
            writer.writerow([s_dict[k], v])
```

```python
# @title Елена Максимова https://stepik.org/lesson/2018079/step/12?discussion=13427249&thread=solutions
import csv

def merge_student_courses(students_file_name, enrollments_file_name):
    with open(students_file_name, encoding='utf-8', ) as f1, open(enrollments_file_name, encoding='utf-8') as f2, open('student_courses.csv', 'w', encoding='utf-8', newline='') as file:
        reader1 = csv.reader(f1, delimiter=':')
        reader2 = csv.reader(f2, delimiter=':')
        writer = csv.writer(file, delimiter=':')
        next(reader1)
        next(reader2)
        d = {}
        for student_id, course in reader2:
            d.setdefault(student_id, []).append(course)
        for student_id, name in reader1:
            if student_id in d:
                d.setdefault(name)
                d[name] = ', '.join(sorted(d[student_id]))
                del d[student_id]
            else:
                d.setdefault(name, 'Пока нет направлений обучения')
        writer.writerow(['name', 'courses'])
        for k, v in d.items():
            writer.writerow([k, v])





```
