# Сводка по жанрам фильмов

```python
import csv

def export_genre_summary(filename):
    with open(filename, encoding='utf-8') as f:
        data = csv.reader(f)
        next(data)
        g = {}
        for row in data:
            g[row[1]] = g.get(row[1], [0, 0, 0])
            g[row[1]][0] += 1
            g[row[1]][1] += float(row[2])
            g[row[1]][2] += float(row[3])

    with open('genre_summary.csv', 'w', encoding='utf-8') as f:
        csv_writer = csv.writer(f)
        csv_writer.writerow(['genre', 'total_movies', 'average_rating', 'average_duration'])
        csv_writer.writerows(sorted([[k, v[0], round(v[2]/v[0], 1), round(v[1]/v[0], 1)] for k, v in g.items()],
                                    key=lambda r: (-r[2], r[0])))

```

```python
# @title Короткая проверка
!wget -nc -q 'https://raw.githubusercontent.com/artemegorof/modules_and_files/refs/heads/main/module_05/lesson_03/step_10/movies1.csv'

export_genre_summary('movies1.csv')

with open('genre_summary.csv', encoding='utf-8') as f:
    print(f.read())

# genre,total_movies,average_rating,average_duration
# Драма,2,8.9,127.0
# Фантастика,2,8.9,158.5
# Боевик,2,8.8,153.5
```
