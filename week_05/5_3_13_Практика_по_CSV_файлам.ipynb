# Ученики и курсы: часть 2

```python
import csv, json

def export_students_to_json(stud_filename, enroll_filename):
    with open(stud_filename, encoding='utf-8') as fs, open(enroll_filename, encoding='utf-8') as fe:
        students = list(csv.reader(fs, delimiter=':'))[1:]
        courses  = list(csv.reader(fe, delimiter=':')) [1:]

        s_dict = {s[0]:s[1] for s in students}
        c_dict = {s[0]:[] for s in students}

        for k, v in courses:
            c_dict[k].append(v)

        courses_list = [{"id": int(k), "name": s_dict[k], "courses": sorted(c_dict[k])} for k in c_dict]

        with open('student_courses.json', 'w', encoding='utf-8') as f:
            json.dump(courses_list, f, indent=2, ensure_ascii=False)
```

```python
# @title Короткая проверка
!wget -nc -q 'https://raw.githubusercontent.com/artemegorof/modules_and_files/refs/heads/main/module_05/lesson_03/step_13/students.csv'
!wget -nc -q 'https://raw.githubusercontent.com/artemegorof/modules_and_files/refs/heads/main/module_05/lesson_03/step_13/enrollments.csv'

export_students_to_json("students.csv", "enrollments.csv")

with open('student_courses.json', encoding='utf-8') as f:
    print(f.read())

# [
#   {
#     "id": 1,
#     "name": "Аня",
#     "courses": [
#       "Git",
#       "Python"
#     ]
#   },
#   {
#     "id": 2,
#     "name": "Борис",
#     "courses": [
#       "SQL"
#     ]
#   },
#   {
#     "id": 3,
#     "name": "Света",
#     "courses": [
#       "Django",
#       "Python",
#       "SQL"
#     ]
#   },
#   {
#     "id": 4,
#     "name": "Тимур",
#     "courses": []
#   },
#   {
#     "id": 5,
#     "name": "Марина",
#     "courses": [
#       "Linux",
#       "Networks"
#     ]
#   }
# ]
```
