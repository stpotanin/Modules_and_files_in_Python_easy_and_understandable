# Каталог и заказы

```python
import csv

def export_order_stats(catalog_filename, orders_filename):
    with open(catalog_filename, encoding='utf-8') as fc, open(orders_filename, encoding='utf-8') as fo:
        reader = csv.DictReader(fc)
        prices = {}
        for row in reader:
            prices[row['id']] = [row['product'], row['price']]

        orders = csv.DictReader(fo)
        order_stats = {k:[0, 0, 0.0] for k in prices.keys()}
        total = 0

        for order in orders:
            quantity = int(order['quantity'])
            price = int(prices[order['product_id']][1])
            revenue = quantity * price

            order_stats[order['product_id']][0] += quantity
            order_stats[order['product_id']][1] += revenue
            total += revenue

        for k in order_stats:
            order_stats[k][2] = round(order_stats[k][1] / total * 100, 1) if total > 0 else 0

    with open('order_stats.csv', 'w', encoding='utf-8', newline='') as f:
        writer = csv.DictWriter(f, fieldnames=['product', 'quantity_sold', 'total_revenue', 'revenue_share'])
        writer.writeheader()
        writer.writerows([{
            'product': prices[k][0],
            'quantity_sold': v[0],
            'total_revenue': v[1],
            'revenue_share': v[2]
        } for k, v in sorted(order_stats.items(), key=lambda item: prices[item[0]][0])])
```

```python
# @title Короткая проверка
!wget -nc -q 'https://raw.githubusercontent.com/artemegorof/modules_and_files/refs/heads/main/module_05/lesson_05/step_15/catalog.csv'
!wget -nc -q 'https://raw.githubusercontent.com/artemegorof/modules_and_files/refs/heads/main/module_05/lesson_05/step_15/orders.csv'

export_order_stats('catalog.csv', 'orders.csv')

with open('order_stats.csv', encoding='utf-8') as f:
    print(f.read())

# product,quantity_sold,total_revenue,revenue_share
# Кофе,1,200,21.1
# Сок,3,450,47.4
# Чай,3,300,31.6
```
