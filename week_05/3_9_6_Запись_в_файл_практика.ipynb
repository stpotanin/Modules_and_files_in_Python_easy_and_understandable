## Сохранение шахматной доски в файл
Ваша задача — сгенерировать шахматную доску размером `n × n` и сохранить её в файл с названием `chess_board.txt`.

Шахматная доска должна представлять собой узор из символов `#` и `.`, чередующихся по горизонтали и вертикали.

Правила построения:

Левая верхняя клетка доски должна быть `#`.

Символы чередуются как по строкам, так и по столбцам — точно как на настоящей шахматной доске.

Каждая строка доски записывается в файл отдельно (с символом переноса строки в конце).

Никакой вывод в консоль делать не нужно — **только запись в файл**.

Для решения задачи напишите функцию `save_board(n)`, которая принимает один аргумент `n` — размер доски (целое число от 1 до 100).

В результате вызова функции `save_board(n)` в текущей директории создаётся файл `chess_board.txt`, в который записана шахматная доска размером `n × n`.

```python
def save_board(n):
    with open('chess_board.txt', 'w') as f:
        s = '#.' * 100
        for _ in range(n):
            f.write(s[:n] + '\n')
            s = s[::-1]
```
