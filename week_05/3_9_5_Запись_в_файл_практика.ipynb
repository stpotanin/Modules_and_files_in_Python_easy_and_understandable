## Статистика по песне
В музыкальном архиве хранится множество текстов песен.

Вам дали задание создать программу, которая должна уметь анализировать их и составлять краткую статистику по каждой песне.

Для этого вам нужно будет написать функцию **`export_song_stats(filename)`**, которая принимает имя текстового файла с песней (например, **`"lyrics.txt"`**), считывает содержимое и создаёт новый файл **`stats.txt`**, в котором для каждой строки указывается количество символов в формате
```
Строка N: X символов
```
                  
где `N` — номер строки (начиная с 1), а `X` — количество символов в этой строке (включая пробелы и знаки препинания).

Так, например, если в файле **`lyrics.txt`** содержится такой текст:
```
Yesterday
All my troubles seemed so far away
Now it looks as though they're here to stay
```
                  
то вызов
```
export_song_stats("lyrics.txt")
```
                  
должен создать файл **`stats.txt`** со следующим содержимым:
```
Строка 1: 9 символов
Строка 2: 34 символов
Строка 3: 43 символов
```
                  
Важно
В тексте могут **встречаться пустые строки** — их длина считается равной нулю.

**Пробелы в конце строки не учитываются** при подсчёте символов.

Все остальные символы — буквы, цифры, знаки препинания, пробелы между словами — **учитываются**.

````python
def export_song_stats(filename):
    with open(filename, 'r', encoding='utf-8') as f, open('stats.txt', 'w', encoding='utf-8') as s:
        [s.write(f'Строка {i+1}: {len(v.strip())} символов\n') for i, v in enumerate(f.readlines())]

        # for i, v in enumerate(f.readlines()):
        #     s.write(f'Строка {i+1}: {len(v.strip())} символов\n')
````
