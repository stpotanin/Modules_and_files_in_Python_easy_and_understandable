## Список покупок
Напишите функцию `save_shopping_list(items)`, которая:

принимает **список строк** `items`, где каждый элемент — это название одного товара;

создает файл с названием `shopping.txt`;

записывает в файл заголовок `Список покупок:` (с двоеточием) в первой строке;

далее записывает **пронумерованный список покупок**, начиная с 1, по одному товару на строку, сохраняя порядок.

К примеру, вызов
```
save_shopping_list(['Хлеб', 'Молоко', 'Яйца'])
```
                  
должен сформировать следующий файл shopping.txt:

Список покупок:
1. Хлеб
2. Молоко
3. Яйца

````python
def save_shopping_list(items):
    with open('shopping.txt', 'w') as f:
        f.write('Список покупок:\n')
        [f.write(str(i + 1) + '. ' + v + '\n') for i, v in enumerate(items)]
````

````python
# @title Проверка
save_shopping_list(['Хлеб', 'Молоко', 'Яйца'])
# Список покупок:
# 1. Хлеб
# 2. Молоко
# 3. Яйца
````
