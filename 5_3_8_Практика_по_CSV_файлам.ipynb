# Объединение двух CSV

```python
# @title Пересечение списков плюс сортировка
import csv

def merge_catalogs(products_filename, prices_filename):
    with open(products_filename, encoding='utf-8') as f1, open(prices_filename, encoding='utf-8') as f2:
        products, prices = list(csv.reader(f1))[1:], list(csv.reader(f2))[1:]
        id = ([x[0] for x in products] and [x[0] for x in prices])
        products = sorted([x for x in products if x[0] in id], key=lambda x: x[0])
        prices = sorted([x for x in prices if x[0] in id], key=lambda x: x[0])
        res = [['id','product','price']] + [[x[0], x[1], y[1]] for x, y in zip(products, prices)]
    with open('merged.csv', 'w', encoding='utf-8') as f:
        writer = csv.writer(f)
        writer.writerows(res)
```

```python
# @title Объединение словарей
import csv

def merge_catalogs(products_filename, prices_filename):
    with open(products_filename, encoding='utf-8') as f1, open(prices_filename, encoding='utf-8') as f2, open('merged.csv', 'w', encoding='utf-8') as f:
        products, prices = dict(csv.reader(f1)), dict(csv.reader(f2))
        writer = csv.writer(f)
        [writer.writerow([key, products[key], prices[key]]) for key in products if key in prices]
```
