# Поиск по названию колонки

```python
import csv

def print_columns(filename, colnames):
    with open(filename, encoding='utf-8') as f:
        reader = csv.reader(f)
        head, body = next(reader), [x for x in reader]
        d = {k: v for v, k in enumerate(head)}

        response = lambda row, d, col: row[d[col]] if col in d else 'N/A'
        if any(col in d for col in colnames):
            [print(*(response(row, d, col) for col in colnames), sep=';') for row in body]
        else:
            print('Указанные колонки не найдены.')
```

```python
# @title Короткая проверка
!wget -nc -q 'https://raw.githubusercontent.com/artemegorof/modules_and_files/refs/heads/main/module_05/lesson_03/step_05/library_books.csv'

print_columns("library_books.csv", ["Автор", "Рейтинг"])

# вызов должен распечатать на экран следующее:

# Джордж Оруэлл;4.8
# Михаил Булгаков;4.9
# Эрих Мария Ремарк;4.7
# Фёдор Достоевский;4.6

print_columns("library_books.csv", ["Автор", "Цена"])

# Джордж Оруэлл;N/A
# Михаил Булгаков;N/A
# Эрих Мария Ремарк;N/A
# Фёдор Достоевский;N/A

print_columns("library_books.csv", ["Цена", "Издатель"])

# Указанные колонки не найдены.
```
