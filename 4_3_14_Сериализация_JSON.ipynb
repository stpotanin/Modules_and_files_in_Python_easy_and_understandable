```python
import json

def export_grouped_comments(comments, filename):
    d = {}
    for c in comments:
        d.setdefault(c['category'], []).append({'user': c['user'], 'text': c['text'], 'likes': c['likes']})
        d[c['category']].sort(key=lambda x: -x['likes'])

    with open(filename, 'w', encoding='utf-8') as f:
        json.dump(d, f, indent=2, ensure_ascii=False)
```
