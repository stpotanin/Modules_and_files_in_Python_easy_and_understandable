## Генератор чеков
Ваша программа помогает магазину формировать понятные и читаемые **чеки покупок**, которые удобно показывать клиенту. Все данные о покупках поступают в программу в виде списка словарей.

Каждый словарь описывает один товар:
```
{
    "название": str,
    "цена": float,         # цена за единицу (может быть вещественной)
    "количество": float    # количество (может быть вещественной)
}
```
                  
Ваша задача — реализовать функцию `generate_receipt(purchases)`, которая:

1. Принимает список покупок purchases, где каждая покупка — это словарь, как указано выше;

2. Создаёт файл с названием `receipt.txt`;

3. Записывает в файл красивый чек по следующему шаблону:
```
==== Чек покупок ====
1. Название (кол-во × цена) = сумма
2. Название (кол-во × цена) = сумма
...
---------------------
ИТОГО: сумма_всего руб
======================
```               
Каждая покупка выводится отдельной строкой, повторяющиеся товары не объединяются.
Все числа (цена, количество, сумма) должны быть округлены до 2 знаков после запятой.
В поле `ИТОГО` выводите общую сумму покупок, также округлённую до 2 знаков после запятой.
Для округления применяйте функцию round.

### Примечания:
* Все числа округлять до 2 знаков после запятой.
* Чек сохраняется **в файл** `receipt.txt`.
* Названия товаров могут быть на русском или английском.

* Между строками **нет пустых строк**.

* Не используйте никаких выравниваний — формат вывода должен быть **точно таким, как в примере**.

````python
def generate_receipt(purchases):

    with open('receipt.txt', 'w', encoding='utf-8') as f:
        total = 0
        f.write('==== Чек покупок ====\n')
        for i, d in enumerate(purchases):
            s = d['количество'] * d['цена']
            total += s
            f.write(f'{i+1}. {d['название']} ({d['количество']:.2f} × {d['цена']:.2f}) = {s:.2f}\n')
        f.write(f'{'-'*21}\nИТОГО: {total:.2f} руб\n{'='*22}')
````

````python
# @title Проверка
purchases = [
    {"название": "Яблоко", "цена": 30.5, "количество": 3},
    {"название": "Хлеб", "цена": 45.1, "количество": 1.5},
    {"название": "Молоко", "цена": 60.44, "количество": 2}
]
generate_receipt(purchases)

with open('receipt.txt', 'r', encoding='utf-8') as f:
    print(f.read())

# ==== Чек покупок ====
# 1. Яблоко (3.00 × 30.50) = 91.50
# 2. Хлеб (1.50 × 45.10) = 67.65
# 3. Молоко (2.00 × 60.44) = 120.88
# ---------------------
# ИТОГО: 280.03 руб
# ======================

purchases = [
    {"название": "Шоколад", "цена": 99.99, "количество": 2},
    {"название": "Сок", "цена": 50.5, "количество": 1.2}
]
generate_receipt(purchases)

with open('receipt.txt', 'r', encoding='utf-8') as f:
    print(f.read())

# ==== Чек покупок ====
# 1. Шоколад (2.00 × 99.99) = 199.98
# 2. Сок (1.20 × 50.50) = 60.60
# ---------------------
# ИТОГО: 260.58 руб
# ======================

purchases = [
    {"название": "Coffee", "цена": 123.45, "количество": 2.5}
]
generate_receipt(purchases)

with open('receipt.txt', 'r', encoding='utf-8') as f:
    print(f.read())

# ==== Чек покупок ====
# 1. Coffee (2.50 × 123.45) = 308.62
# ---------------------
# ИТОГО: 308.62 руб
# ======================
````
