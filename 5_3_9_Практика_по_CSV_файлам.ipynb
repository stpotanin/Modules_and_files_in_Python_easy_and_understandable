# Статистика по отзывам

```python
import json
import csv
from collections import defaultdict

def export_user_stats(filename):
    with open(filename, encoding='utf-8') as f:
        data = json.load(f)
        stats_agg = defaultdict(lambda: [0, 0])
        for d in data:
            stats_agg[d['user']][0] += 1
            stats_agg[d['user']][1] += d['rating']
        stats = dict(sorted(stats_agg.items()))

    with open('user_stats.csv', 'w', encoding='utf-8', newline='') as f:
        writer = csv.writer(f)
        writer.writerow(['user', 'total_reviews', 'average_rating'])
        for k, v in stats.items():
            average_rating = v[1] / v[0] if v[0] > 0 else 0
            writer.writerow([k, v[0], round(average_rating, 1)])
```

```python
# @title Короткая проверка

export_user_stats('reviews.json')

with open('user_stats.csv', encoding='utf-8') as f:
    print(f.read())

# user,total_reviews,average_rating
# alice,2,4.5
# bob,1,3.0
```
