```python
def longest_palindrome(filename):
    max_len = 0
    with open(filename, 'r') as f:
        text = f.read().split()
        for word in text:
            if word.lower() == word[::-1].lower() and len(word) > max_len:
                max_len = len(word)
                longest_word = word

        return longest_word

print(longest_palindrome('mixed_case.txt'))
```
