```python
import json

def decode_text(file_name, alpha_name):
    with open(file_name, 'r') as f, open(alpha_name) as a:
        text = f.read()                 # читаем весь текст как строку
        cifer = json.load(a)

    # декодируем каждый символ
    decoded_chars = []
    for ch in text:
        if ch in cifer:
            decoded_chars.append(cifer[ch])
        else:
            decoded_chars.append(ch)

    result = ''.join(decoded_chars)

    with open('decoded.txt', 'w') as d:
        d.write(result)
```
