# Финансовый отчёт по клиентам

```python
import xml.etree.ElementTree as ET

def export_financial_report(filename, transactions):
    root = ET.Element('report')
    d = {}
    for t in transactions:
        key = (t['client'], t['currency'])
        d[key] = d.get(key, [0, []])
        d[key][0] += t['amount']
        d[key][1].append(t['amount'])
    for k, v in d.items():
        client = ET.SubElement(root, 'client', name=k[0])
        ET.SubElement(client, 'total').text = str(v[0])
        ET.SubElement(client, 'currency').text = k[1]
        transactions = ET.SubElement(client, 'transactions')
        for t in v[1]:
            ET.SubElement(transactions, 'transaction').text = str(t)

    tree = ET.ElementTree(root)
    ET.indent(tree)
    tree.write(filename, encoding="utf-8")
```

```python
# @title Краткая проверка
clients = [
    {"client": "Warren Buffett", "amount": 120, "currency": "USD"},
    {"client": "Warren Buffett", "amount": 80, "currency": "RUB"}, # USD
    {"client": "Tony Stark", "amount": 200, "currency": "EUR"},
]

export_financial_report('report1.xml', clients)

with open('report1.xml', 'r', encoding='utf-8') as f:
    print(f.read())
# <report>
#   <client name="Warren Buffett">
#     <total>200</total>
#     <currency>USD</currency>
#     <transactions>
#       <transaction>120</transaction>
#       <transaction>80</transaction>
#     </transactions>
#   </client>
#   <client name="Tony Stark">
#     <total>200</total>
#     <currency>EUR</currency>
#     <transactions>
#       <transaction>200</transaction>
#     </transactions>
#   </client>
# </report>

```

```python
import xml.etree.ElementTree as ET


def export_financial_report(filename, transactions):
    result_dict = {}

    for item in transactions:
        if item['client'] not in result_dict:
            result_dict.setdefault(item['client'], {'amount': [], 'currency': item['currency']})
        result_dict[item['client']]['amount'].append(item['amount'])

    root = ET.Element('report')

    for k, v in result_dict.items():
        client = ET.SubElement(root, 'client')
        client.set('name', k)

        total = ET.SubElement(client, 'total')
        total.text = str(sum(v['amount']))
        currency = ET.SubElement(client, 'currency')
        currency.text = v['currency']
        transactions = ET.SubElement(client, 'transactions')

        for nums in v['amount']:
            transaction = ET.SubElement(transactions, 'transaction')
            transaction.text = str(nums)

    tree = ET.ElementTree(root)
    ET.indent(tree)
    tree.write(filename, encoding='utf-8')
```
