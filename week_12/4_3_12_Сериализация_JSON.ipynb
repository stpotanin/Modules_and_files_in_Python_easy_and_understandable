```python
import json

def save_profile(compact=False, sort=False, **kwargs):
    with open('profile.json', 'w', encoding='utf-8') as profile:
        json.dump(dict(sorted(kwargs.items())) if sort else kwargs,
                  profile,
                  ensure_ascii=False,
                  indent=None if compact else 2
        )
```

```python
# @title Проверка
save_profile(
    name="Артем",
    age=35,
    city="Москва",
    hobbies=["теннис", "чтение", "программирование"]
)

with open('profile.json', 'r', encoding='utf-8') as f:
    print(f.read())

# {
#   "name": "Артем",
#   "age": 35,
#   "city": "Москва",
#   "hobbies": [
#     "теннис",
#     "чтение",
#     "программирование"
#   ]
# }
print('------------------------------')

save_profile(
    name="Андрей",
    age=40,
    occupation="Data Engineer",
    country="Россия",
    compact=True,
    sort=True
)

with open('profile.json', 'r', encoding='utf-8') as f:
    print(f.read())

# {"age": 40, "country": "Россия", "name": "Андрей", "occupation": "Data Engineer"}
print('------------------------------')

save_profile(
    name="Мария",
    age=28,
    city="Санкт-Петербург",
    subscribed=None,
    sort=True
)

with open('profile.json', 'r', encoding='utf-8') as f:
    print(f.read())

# {
#   "age": 28,
#   "city": "Санкт-Петербург",
#   "name": "Мария",
#   "subscribed": null
# }
print('------------------------------')

save_profile(
    compact=True,
    name="Мария",
    age=28,
    city="Санкт-Петербург",
    subscribed=None,
    sort=False,
)

with open('profile.json', 'r', encoding='utf-8') as f:
    print(f.read())

# {"name": "Мария", "age": 28, "city": "Санкт-Петербург", "subscribed": null}
```
