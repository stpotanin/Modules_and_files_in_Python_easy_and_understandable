# Самый загруженный маршрут

```python
import xml.etree.ElementTree as ET

def get_heaviest_route(filename):
    root = ET.parse(filename).getroot()
    d = {}
    for route in root.findall('route'):
        f, t = route.find('from').text, route.find('to').text
        weight = int(route.find('weight').text)
        d[(f, t)] = d.get((f, t), 0) + weight
        max_key = max(d, key=d.get)

    return *max_key, d[max_key]
```

```python
# @title Краткая проверка
!wget -nc -q 'https://raw.githubusercontent.com/artemegorof/modules_and_files/refs/heads/main/module_07/lesson_04/step_08/routes_1.xml'

print(get_heaviest_route("routes_1.xml"))
# ('Berlin', 'Paris', 2000)
```
