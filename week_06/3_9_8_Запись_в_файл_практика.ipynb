## Битва языков программирования
В сообществе **«Код без кофеина»** прошёл жаркий опрос:

> Какой язык программирования достоин звания "Короля синтаксиса"?

Теперь тебе нужно подвести итоги и подготовить их к публикации на сайте.

Напиши функцию `export_poll_results(results)`, которая принимает словарь с результатами голосования, например:
```
{
    "Python": 38,
    "JavaScript": 25,
    "C++": 12,
    "Java": 17
}
```      
и сохраняет данные в файл `results.txt` в следующем виде:
```
Результаты опроса:
1. Python — 38 голосов
2. JavaScript — 25 голосов
3. Java — 17 голосов
4. C++ — 12 голосов
```
        
**Требования:**

* Имя файла всегда одинаковое — `results.txt`.

* Результаты сортируются **по убыванию количества голосов** — от победителя к аутсайдеру.

* Формат текста должен точно совпадать с примером (включая нумерацию, двоеточие и длинное тире).

* Гарантируется, что количество голосов у всех языков разное.

````python
def export_poll_results(results):
    with open('results.txt', 'w', encoding='utf-8') as f:
        f.write('Результаты опроса:\n')
        for k, v in enumerate(sorted(results.items(), key=lambda i: -i[1])):
            f.write(f'{k+1}. {v[0]} — {v[1]} голосов\n')
````

````python
# @title Проверка
results = {
    "Python": 38,
    "JavaScript": 25,
    "C++": 12,
    "Java": 17
}
export_poll_results(results)

with open('results.txt', 'r', encoding='utf-8') as f:
    print(f.read())

# Результаты опроса:
# 1. Python — 38 голосов
# 2. JavaScript — 25 голосов
# 3. Java — 17 голосов
# 4. C++ — 12 голосов

results = {
    "Swift": 8,
    "Kotlin": 2,
    "Java": 15,
    "Ruby": 5,
    "Pascal": 3,
    "Basic": 1,
}
export_poll_results(results)

with open('results.txt', 'r', encoding='utf-8') as f:
    print(f.read())

# Результаты опроса:
# 1. Java — 15 голосов
# 2. Swift — 8 голосов
# 3. Ruby — 5 голосов
# 4. Pascal — 3 голосов
# 5. Kotlin — 2 голосов
# 6. Basic — 1 голосов

results = {
    "Rust": 40,
    "Go": 60,
    "Python": 78,
    "C": 10
}
export_poll_results(results)

with open('results.txt', 'r', encoding='utf-8') as f:
    print(f.read())

# Результаты опроса:
# 1. Python — 78 голосов
# 2. Go — 60 голосов
# 3. Rust — 40 голосов
# 4. C — 10 голосов
````
