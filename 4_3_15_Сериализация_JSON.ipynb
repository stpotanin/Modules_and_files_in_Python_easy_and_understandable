```python
import json

def export_comments_by_category(comments):
    d = {}
    for c in comments:
        d.setdefault(c['category'], []).append({'user': c['user'], 'text': c['text'], 'likes': c['likes']})
        d[c['category']].sort(key=lambda x: -x['likes'])

    for k, v in d.items():
        with open(k.lower() + '_comments.json', 'w', encoding='utf-8') as f:
            json.dump(v, f, indent=2, ensure_ascii=False)
```
