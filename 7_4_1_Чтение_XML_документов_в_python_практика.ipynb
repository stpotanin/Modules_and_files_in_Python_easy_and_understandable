# Получение корневого тега

```python
def get_root_tag(filename):
    import xml.etree.ElementTree as ET

    tree = ET.parse(filename)
    root = tree.getroot()
    return root.tag
```

```python
import xml.etree.ElementTree as ET

get_root_tag = lambda filename: ET.parse(filename).getroot().tag
```
