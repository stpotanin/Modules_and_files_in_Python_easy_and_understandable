# Экспорт матрицы в CSV

```python
def save_matrix_to_csv(matrix, file_name):
    with open(file_name, 'w') as f: f.write('\n'.join(','.join(str(x) for x in row) for row in matrix))
```

```python
# @title Проверка
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

file_name = "matrix.csv"

save_matrix_to_csv(matrix, file_name)

with open(file_name, encoding='utf-8') as f:
    print(f.read())

# 1,2,3
# 4,5,6
# 7,8,9
print()

matrix = [
    [i, i * 10, i * 100, i * 1000] for i in range(1, 11)
]

file_name = "matrix_table.csv"

save_matrix_to_csv(matrix, file_name)

with open(file_name, encoding='utf-8') as f:
    print(f.read())

# 1,10,100,1000
# 2,20,200,2000
# 3,30,300,3000
# 4,40,400,4000
# 5,50,500,5000
# 6,60,600,6000
# 7,70,700,7000
# 8,80,800,8000
# 9,90,900,9000
# 10,100,1000,10000
print()

matrix = [[1.1, 2.25, 3.75],
          [4.0, 5.5, 6.25]]

file_name = "matrix_float.csv"

save_matrix_to_csv(matrix, file_name)

with open(file_name, encoding='utf-8') as f:
    print(f.read())

# 1.1,2.25,3.75
# 4.0,5.5,6.25

```
