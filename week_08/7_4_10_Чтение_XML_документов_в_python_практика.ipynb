# Продавцы с высоким рейтингом

```python
import xml.etree.ElementTree as ET

def top_sellers(filename):
    root = ET.parse(filename).getroot()
    d = {}
    for seller in root.findall('seller'):
        sku = seller.attrib['id']
        d[sku] = d.get(sku, [0, 0])
        for review in seller.findall('.//review'):
            d[sku][0] += int(review.attrib['rating'])
            d[sku][1] += 1

    d = {sku : [d[sku][0] / d[sku][1], d[sku][1]] if d[sku][1] else 0
         for sku in d if d[sku][0] / d[sku][1] >= 4.5 and d[sku][1] >= 10}

    return [(k, round(v[0], 2), v[1]) for k, v in d.items()]
```

```python
# @title Краткая проверка
!wget -nc -q 'https://raw.githubusercontent.com/artemegorof/modules_and_files/refs/heads/main/module_07/lesson_04/step_10/market_1.xml'
!wget -nc -q 'https://raw.githubusercontent.com/artemegorof/modules_and_files/refs/heads/main/module_07/lesson_04/step_10/market_2.xml'

print(top_sellers("market_1.xml"))
# [('s1', 4.7, 10)]
print()

print(top_sellers("market_2.xml"))
# [('alpha', 5.0, 10)]
```
