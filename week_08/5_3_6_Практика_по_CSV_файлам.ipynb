# Список фильмов с описанием

```python
import csv

def print_movies(filename):
    with open(filename, encoding='utf-8') as f:
        # 1. Используйте модуль csv и csv.reader для построчного чтения файла.
        reader = csv.reader(f)

        # 2. Пропустите строку с заголовками.
        next(reader)

        # 3. Для каждой строки распакуйте значения в четыре переменные: title, genre, duration, rating.
        title, genre, duration, rating = [], [], [], []
        for row in reader:
            title.append(row[0])
            genre.append(row[1])
            duration.append(row[2])
            rating.append(row[3])

        # 4. Выведите информацию о фильме в формате:
        for t, g, d, r in zip(title, genre, duration, rating):
            print(f'Фильм: {t} | Жанр: {g} | Длительность: {d} мин | Рейтинг: {r}')
```
