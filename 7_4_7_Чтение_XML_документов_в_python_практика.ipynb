# Отели и бронирования

```python
import xml.etree.ElementTree as ET

def calculate_revenue(filename):
    root = ET.parse(filename).getroot()
    d = {}
    for elem in root.findall('booking'):
        hotel = elem.find('hotel').text
        nights = int(elem.find('nights').text)
        price = int(elem.find('price').text)
        d[hotel] = d.get(hotel, 0) + nights * price
    return d
```

```python
# @title Краткая проверка
!wget -nc -q 'https://raw.githubusercontent.com/artemegorof/modules_and_files/refs/heads/main/module_07/lesson_04/step_07/bookings_1.xml'

print(calculate_revenue("bookings_1.xml"))
# {'Grand Hotel': 480, 'City Inn': 160}
```
