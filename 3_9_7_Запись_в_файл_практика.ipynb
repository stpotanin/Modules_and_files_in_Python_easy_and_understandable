## Генерация файла с числами
Напишите функцию `create_file_with_numbers(n)`, которая принимает одно целое положительное число `n` и создает файл с именем:
```
range_<n>.txt
```
                  
Функция должна в этот файл записать все целые числа от 1 до n включительно. Каждое число — с новой строки. Обязательно добавляйте символ переноса строки после каждого числа, включая последнее.

Пример: при вызове
```
create_file_with_numbers(5)
```
                  
должен создаться файл `range_5.txt` с содержимым
```
1
2
3
4
5
```
                  
Примечания:

Гарантируется, что `n` — положительное целое число.

Все числа записываются **строго по одному на строку**.

Последняя строка также должна **заканчиваться переносом строки** `\n`, иначе некоторые тесты могут не пройти.

````python
def create_file_with_numbers(n):
    with open(f'range_{n}.txt', 'w', encoding='utf-8') as f:
        [f.write(f'{i}\n') for i in range(1, n+1)]
````
