## Морской бой: обновление карты
В распоряжении вашей программы два файла:

**`field_file`** — имя файла, содержащего поле морского боя размером **10×10**. Каждая строка — ряд поля. Внутри могут быть символы:

* ~ — вода,

* ■ — часть корабля.

**`shots_file`** — имя файла, содержащего координаты выстрелов. Каждая строка — одна координата в формате «буква+цифра» (например: `B4`, `J10`).

🛠 Ваша задача — написать функцию `mark_hits(field_file, shots_file)`, которая должна:

* Открыть файл `field.txt` и загрузить поле.

* Обработать все выстрелы из файла `shots.txt`.

* Отметить в копии поля:

        * `X`, если выстрел попал по кораблю (было `■`);

        * `•`, если выстрел пришёлся по воде (было `~`);

* Все остальные клетки оставить без изменений.

Сохранить обновлённое поле в файл `updated_field.txt`.

Например, у файле `field.txt` находится следующее поле
```
~ ~ ~ ~ ~ ~ ~ ~ ~ ~
~ ■ ~ ~ ~ ~ ~ ~ ~ ~
~ ■ ~ ~ ~ ~ ~ ~ ~ ~
~ ~ ~ ~ ~ ~ ~ ~ ~ ~
~ ~ ~ ■ ■ ~ ~ ~ ~ ~
~ ~ ~ ~ ~ ~ ~ ~ ~ ~
~ ~ ~ ~ ~ ~ ~ ~ ~ ~
~ ~ ~ ~ ~ ~ ~ ~ ~ ~
~ ~ ~ ~ ~ ~ ~ ~ ~ ~
~ ~ ~ ~ ~ ~ ~ ~ ~ ~
```
                  
Игрок наносит следующие выстрелы, они лежат в файле `shots.txt`:
```
B2
B3
D5
A1
E5
J10
```
                  
Тогда в итоговом файле `updated_field.txt` вы должны сохранить следующее поле:
```
• ~ ~ ~ ~ ~ ~ ~ ~ ~
~ X ~ ~ ~ ~ ~ ~ ~ ~
~ X ~ ~ ~ ~ ~ ~ ~ ~
~ ~ ~ ~ ~ ~ ~ ~ ~ ~
~ ~ ~ X X ~ ~ ~ ~ ~
~ ~ ~ ~ ~ ~ ~ ~ ~ ~
~ ~ ~ ~ ~ ~ ~ ~ ~ ~
~ ~ ~ ~ ~ ~ ~ ~ ~ ~
~ ~ ~ ~ ~ ~ ~ ~ ~ ~
~ ~ ~ ~ ~ ~ ~ ~ ~ •
```
                  
### 📦 Гарантируется:
* Поле всегда размером 10×10.
* Координаты во входном файле всегда валидны (от A1 до J10).
* Файл `field.txt` всегда содержит 10 строк по 10 символов, разделённых пробелами.
* Все символы в поле: `~`, `■`, `X`, `•`. Других символов в тестовых данных не будет

````python
def mark_hits(field_file, shots_file):
    rule = {'■':'X', '~':'•'}
    with open(field_file, 'r', encoding='utf-8') as f_field, open(shots_file, 'r', encoding='utf-8') as f_shots:
        field = [line.split() for line in f_field.read().splitlines()]
        shots = f_shots.read().splitlines()

        for shot in shots:
            row_char = shot[0]
            col_num_str = shot[1:]

            row_idx = int(col_num_str) - 1   # '1' -> 0, '2' -> 1, и т.д.
            col_idx = ord(row_char) - ord('A') # 'A' -> 0, 'B' -> 1, и т.д.

            current_char = field[row_idx][col_idx]
            # всё-таки подстрахуемся насчёт входных символов
            if current_char in rule:
                field[row_idx][col_idx] = rule[current_char]

    with open('updated_field.txt', 'w', encoding='utf-8') as f_out:
        for i, row in enumerate(field):
            f_out.write(' '.join(row))
            if i < len(field) - 1:
                f_out.write('\n')
````
