# Построчное чтение CSV-файла

```python
import csv

def read_csv(filename):
    with open(filename, encoding='utf-8') as f:
        [print(row) for row in csv.reader(f)]
```

```python
# @title Короткая проверка
!wget -nc -q 'https://raw.githubusercontent.com/artemegorof/modules_and_files/refs/heads/main/module_05/lesson_03/step_01/books.csv'

read_csv('books.csv')

# вызов read_csv(books.csv)  должен распечатать на экран следующее:

# ['ID', 'Название', 'Автор', 'Жанр', 'Страниц', 'Рейтинг']
# ['1', '1984', 'Джордж Оруэлл', 'Фантастика', '328', '4.8']
# ['2', 'Мастер и Маргарита', 'Михаил Булгаков', 'Роман', '480', '4.9']
# ['3', 'Собачье сердце', 'Михаил Булгаков', 'Сатира', '176', '4.8']
```
