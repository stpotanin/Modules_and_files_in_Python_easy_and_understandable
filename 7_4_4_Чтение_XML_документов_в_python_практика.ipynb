# Активные курсы

```python
import xml.etree.ElementTree as ET

def get_active_courses(filename):
    root = ET.parse(filename).getroot()
    [print(f'{course.find('title').text} — {course.find('duration').text}')
    for course in root.iter('course') if course.attrib['active'].lower() == 'true']
```
